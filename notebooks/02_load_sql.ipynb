```python
import pandas as pd
from sqlalchemy import create_engine

train = pd.read_csv("../data/raw/train.csv", dtype={"StateHoliday": str})
store = pd.read_csv("../data/raw/store.csv")

engine = create_engine("sqlite:///../data/rossmann.db")

train.to_sql("sales", engine, if_exists="replace", index=False)
store.to_sql("stores", engine, if_exists="replace", index=False)

print("Done")
```

```python
pd.read_sql("SELECT * FROM sales LIMIT 5", engine)
```

```python
pd.read_sql("SELECT * FROM stores LIMIT 5", engine)
```

```python
pd.read_sql("SELECT COUNT(*) FROM sales", engine)
```
